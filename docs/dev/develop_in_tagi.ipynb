#### **Start Developing in TAGI**

For developing inside the robot, its important to have access to it.

> **How to connect to TAGI**

**Step 1** ::: Connect a mouse and keyboard with the USB dongle available.

**Step 2** ::: Use the keyboard shortcut ``CTRL+ALT+T`` to open the terminal.

**Step 3** ::: Connect the robot to the WiFi with the command bellow.

```python
nmcli dev wifi connect <SSID> password <password>
```

Notes:

- This can also be achieved on robot's Wifi Connect mode. Check [Behaviour Usage](/home/idmind/elmo-v2/docs/user/behaviour_usage.ipynb) to learn how to do it. 

- This can also be achieved by attaching an Ethernet cable to the robot.

**Step 4** ::: Get the robot's IP address with the command bellow. This might be a useful information for the next steps.

```python
hostname -I
```

> **Running the Development APP**

The development app is where all the robot's hardware can be tested.

**Step 1** ::: To use the app, get remote access to the robot through SSH. On the ``<IP>`` area, swap it with the robot's IP address.

```python
ssh idmind@<IP>
```

###### Note: Since all the robots have the same username and the same hostname, the command ``ssh idmind@tagi.local`` should work.

**Step 2** ::: Get inside the app's source directory.

```python
cd ~/elmo-v2/app/src
```

**Step 3** ::: Run ``app.py``.

```python
uv run app.py
```

Important note ::: This app only runs on Windows.

If the wish is to run on Linux follow the steps bellow:

**Step 1** ::: Open ``pyproject.toml``. It looks something like this:

```python
[project]
name = "TAGI-App"
version = "1.3.0"
description = "Companion App for TAGI. It can be used to test all the hardware features of TAGI and also provides a user interface for users to interact with TAGI."
readme = "README.md"
requires-python = ">3.10, <3.14"
dependencies = [
    "requests>=2.32.5",
    "PyQt5>=5.15.0",
    "netifaces-plus>=0.12.5"
]
[tool.uv]
    required-environments = [
        "sys_platform == 'win32' and platform_machine == 'AMD64'",
]
```

**Step 2** ::: Do the following changes:
- Remove [tool.uv] section;
- Possibly change the dependencies and the Python version. This may require actually engaging and changing the app's code.
